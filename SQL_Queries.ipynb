## Setup
Correct the error and run this cell first to import required libraries and define the helper function.

```python
import sqlite3
import pandas as pd
from tabulate import tabulate

conn = sqlite3.connect("rentio.db")
cursor = conn.cursor()

def run_sql(query):
    return pd.read_sql_query(query, conn)
```

```python
cursor.executescript("""
DROP TABLE IF EXISTS category;
DROP TABLE IF EXISTS film;
DROP TABLE IF EXISTS store;
DROP TABLE IF EXISTS staff;
DROP TABLE IF EXISTS film_category;

CREATE TABLE category (
    category_id INTEGER PRIMARY KEY,
    name TEXT,
    last_update TEXT
);

CREATE TABLE film (
    film_id INTEGER PRIMARY KEY,
    title TEXT,
    length INTEGER,
    release_year INTEGER,
    rating TEXT
);

CREATE TABLE store (
    store_id INTEGER PRIMARY KEY,
    manager_staff_id INTEGER,
    address_id INTEGER,
    last_update TEXT
);

CREATE TABLE staff (
    staff_id INTEGER PRIMARY KEY,
    first_name TEXT,
    last_name TEXT,
    email TEXT,
    active INTEGER,
    address_id INTEGER
);

CREATE TABLE film_category (
    film_id INTEGER,
    category_id INTEGER
);
""")

conn.commit() # It saves the DB from RAM to HardDisk
```

## Question 4: Aggregate Functions
Calculate the average length of all films. Rename the column as `average_film_length`.

```python
# TODO
run_sql("""
SELECT AVG(length) AS average_film_length FROM film
""")
```

## Question 5: Joining Tables
Display each film title along with its category name.

```python
# TODO
run_sql("""
SELECT film.title, category.name AS category_name
FROM film
JOIN film_category ON film.film_id = film_category.film_id
JOIN category ON film_category.category_id = category.category_id
""")
```

## Question 6: GROUP BY and COUNT
Show the number of films in each category, sorted from highest to lowest count.

```python
# TODO
run_sql("""
SELECT category.name AS category_name, COUNT(film.film_id) AS film_count
FROM category
JOIN film_category ON category.category_id = film_category.category_id
JOIN film ON film_category.film_id = film.film_id
GROUP BY category.name
ORDER BY film_count DESC
""")
```

## Question 7: Handling Duplicate Data
Display distinct category names only.

```python
# TODO
run_sql("""
SELECT DISTINCT category.name AS category_name
FROM category
JOIN film_category ON category.category_id = film_category.category_id
""")
```

## Question 8: Conditional Logic
Display film title and classify films as `Long Film` or `Short Film`.

```python
# TODO
run_sql("""
SELECT title,
CASE
    WHEN length > 120 THEN 'Long Film'
    ELSE 'Short Film'
END AS film_type
FROM film
""")
```

## Question 9: Multi-Table Query
Display Store ID, Manager’s full name, and Manager’s email.

```python
# TODO
run_sql("""
SELECT store.store_id, staff.first_name || ' ' || staff.last_name AS manager_name, staff.email
FROM store
JOIN staff ON store.manager_staff_id = staff.staff_id
""")
```

## Question 10: Data Validation Query
Identify categories that have no films assigned to them.

```python
# TODO
run_sql("""
SELECT category.name AS category_name
FROM category
LEFT JOIN film_category ON category.category_id = film_category.category_id
WHERE film_category.film_id IS NULL
""")
```
